```python
df_long["YearStart"] = df_long["Year"].str[:4].astype(int)
```

```python
df_long[["Year", "YearStart"]].head()
```

```python
df_long = df_long.sort_values(
    ["Province", "Station", "Category", "YearStart"]
).reset_index(drop=True)
```

```python
df_long["Lag1"] = (
    df_long.groupby(["Province", "Station", "Category"])["CrimeCount"]
    .shift(1)
)
```

```python
df_long["Lag2"] = (
    df_long.groupby(["Province", "Station", "Category"])["CrimeCount"]
    .shift(2)
)
```

```python
df_long["TargetNextYear"] = (
    df_long.groupby(["Province", "Station", "Category"])["CrimeCount"]
    .shift(-1)
)
```

```python
df_long[
    [
        "Province",
        "Station",
        "Category",
        "Year",
        "CrimeCount",
        "Lag1",
        "Lag2",
        "TargetNextYear"
    ]
].head(15)
```

```python
df_long[["Lag1", "Lag2", "TargetNextYear"]].isnull().sum()
```

```python
ml_data = df_long.dropna(
    subset=["CrimeCount", "Lag1", "Lag2", "TargetNextYear"]
).copy()
```

```python
ml_data.shape
```

```python
ml_data[
    ["Province", "Station", "Category",
     "Year", "CrimeCount", "Lag1", "Lag2", "TargetNextYear"]
].head(10)
```

```python
sorted(ml_data["YearStart"].unique())
```

```python
train = ml_data[ml_data["YearStart"] <= 2012].copy()

validation = ml_data[
    ml_data["YearStart"] == 2013
].copy()

test = ml_data[
    ml_data["YearStart"] == 2014
].copy()
```

```python
print("Train:", train.shape)
print("Validation:", validation.shape)
print("Test:", test.shape)
```

```python
print("Train years:", sorted(train["YearStart"].unique()))
print("Validation years:", sorted(validation["YearStart"].unique()))
print("Test years:", sorted(test["YearStart"].unique()))
```

```python
features = ["YearStart", "CrimeCount", "Lag1", "Lag2"]
```

```python
target = "TargetNextYear"
```

```python
X_train = train[features]
y_train = train[target]

X_validation = validation[features]
y_validation = validation[target]

X_test = test[features]
y_test = test[target]
```

```python
baseline_pred = X_test["CrimeCount"]
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
```

```python
baseline_mae = mean_absolute_error(y_test, baseline_pred)

baseline_rmse = np.sqrt(
    mean_squared_error(y_test, baseline_pred)
)

baseline_r2 = r2_score(y_test, baseline_pred)
```

```python
print("Baseline MAE:", baseline_mae)
print("Baseline RMSE:", baseline_rmse)
print("Baseline R²:", baseline_r2)
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)
```

```python
validation_pred = rf_model.predict(X_validation)
```

```python
validation_mae = mean_absolute_error(
    y_validation,
    validation_pred
)

validation_rmse = np.sqrt(
    mean_squared_error(
        y_validation,
        validation_pred
    )
)

validation_r2 = r2_score(
    y_validation,
    validation_pred
)
```

```python
print("Validation MAE:", validation_mae)
print("Validation RMSE:", validation_rmse)
print("Validation R²:", validation_r2)
```

```python
test_pred = rf_model.predict(X_test)
```

```python
test_mae = mean_absolute_error(y_test, test_pred)

test_rmse = np.sqrt(
    mean_squared_error(y_test, test_pred)
)

test_r2 = r2_score(y_test, test_pred)
```

```python
print("Test MAE:", test_mae)
print("Test RMSE:", test_rmse)
print("Test R²:", test_r2)
```

```python
results = pd.DataFrame({
    "Model": ["Persistence Baseline", "Random Forest"],
    "MAE": [baseline_mae, test_mae],
    "RMSE": [baseline_rmse, test_rmse],
    "R2": [baseline_r2, test_r2]
})

results
```

```python
train_pred = rf_model.predict(X_train)
```

```python
train_mae = mean_absolute_error(y_train, train_pred)

train_rmse = np.sqrt(
    mean_squared_error(y_train, train_pred)
)

train_r2 = r2_score(
    y_train,
    train_pred
)
```

```python
("TRAIN")
print("MAE:", train_mae)
print("RMSE:", train_rmse)
print("R²:", train_r2)

print("\nVALIDATION")
print("MAE:", validation_mae)
print("RMSE:", validation_rmse)
print("R²:", validation_r2)

print("\nTEST")
print("MAE:", test_mae)
print("RMSE:", test_rmse)
print("R²:", test_r2)print
```

```python
model_performance = pd.DataFrame({
    "Dataset": ["Train", "Validation", "Test"],
    "MAE": [train_mae, validation_mae, test_mae],
    "RMSE": [train_rmse, validation_rmse, test_rmse],
    "R²": [train_r2, validation_r2, test_r2]
})

model_performance
```

```python
rf_tuned = RandomForestRegressor(
    n_estimators=100,
    max_depth=10,
    min_samples_leaf=2,
    max_features="sqrt",
    random_state=42
)
```

```python
rf_tuned.fit(X_train, y_train)
```

```python
tuned_train_pred = rf_tuned.predict(X_train)

tuned_validation_pred = rf_tuned.predict(X_validation)

tuned_test_pred = rf_tuned.predict(X_test)
```

```python
tuned_train_mae = mean_absolute_error(y_train, tuned_train_pred)
tuned_train_rmse = np.sqrt(mean_squared_error(y_train, tuned_train_pred))
tuned_train_r2 = r2_score(y_train, tuned_train_pred)

tuned_validation_mae = mean_absolute_error(y_validation, tuned_validation_pred)
tuned_validation_rmse = np.sqrt(mean_squared_error(y_validation, tuned_validation_pred))
tuned_validation_r2 = r2_score(y_validation, tuned_validation_pred)

tuned_test_mae = mean_absolute_error(y_test, tuned_test_pred)
tuned_test_rmse = np.sqrt(mean_squared_error(y_test, tuned_test_pred))
tuned_test_r2 = r2_score(y_test, tuned_test_pred)
```

```python
print("TUNED MODEL")

print("\nTRAIN")
print("MAE:", tuned_train_mae)
print("RMSE:", tuned_train_rmse)
print("R²:", tuned_train_r2)

print("\nVALIDATION")
print("MAE:", tuned_validation_mae)
print("RMSE:", tuned_validation_rmse)
print("R²:", tuned_validation_r2)

print("\nTEST")
print("MAE:", tuned_test_mae)
print("RMSE:", tuned_test_rmse)
print("R²:", tuned_test_r2)
```

```python
model_comparison = pd.DataFrame({
    "Model": ["Original Random Forest", "Tuned Random Forest"],
    "Test MAE": [test_mae, tuned_test_mae],
    "Test RMSE": [test_rmse, tuned_test_rmse],
    "Test R²": [test_r2, tuned_test_r2]
})

model_comparison
```

```python
print("Original R² gap:", train_r2 - validation_r2)
print("Tuned R² gap:", tuned_train_r2 - tuned_validation_r2)
```

```python
print("Original RMSE gap:", validation_rmse - train_rmse)
print("Tuned RMSE gap:", tuned_validation_rmse - tuned_train_rmse)
```

```python
if tuned_validation_mae < validation_mae:
    final_model = rf_tuned
    final_model_name = "Tuned Random Forest"
else:
    final_model = rf_model
    final_model_name = "Original Random Forest"

print("Selected model:", final_model_name)
```

```python
final_test_pred = final_model.predict(X_test)

final_mae = mean_absolute_error(y_test, final_test_pred)
final_rmse = np.sqrt(mean_squared_error(y_test, final_test_pred))
final_r2 = r2_score(y_test, final_test_pred)

print("Final Model:", final_model_name)
print("Test MAE:", final_mae)
print("Test RMSE:", final_rmse)
print("Test R²:", final_r2)
```

```python
from statsmodels.tsa.holtwinters import ExponentialSmoothing
```

```python
ts_model = ExponentialSmoothing(
    ts_train["CrimeCount"],
    trend="add",
    seasonal=None
)
```

```python
ts_fit = ts_model.fit()
```

```python
ts_forecast = ts_fit.forecast(
    steps=len(ts_test)
)

ts_forecast
```

```python
ts_mae = mean_absolute_error(
    ts_test["CrimeCount"],
    ts_forecast
)

ts_rmse = np.sqrt(
    mean_squared_error(
        ts_test["CrimeCount"],
        ts_forecast
    )
)

print("Exponential Smoothing MAE:", ts_mae)
print("Exponential Smoothing RMSE:", ts_rmse)
```
